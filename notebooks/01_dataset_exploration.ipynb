```python
date_columns = [
    "order date (DateOrders)",
    "shipping date (DateOrders)"
]

for column in date_columns:
    df[column] = pd.to_datetime(
        df[column],
        format="%m/%d/%Y %H:%M",
        errors="coerce"
    )

print(df[date_columns].dtypes)

display(df[date_columns].head())
```

```python
for column in date_columns:
    print(f"\nColumn: {column}")
    print("Missing dates:", df[column].isna().sum())
    print("Earliest date:", df[column].min())
    print("Latest date:", df[column].max())
```

```python
df["Calculated Shipping Days"] = (
    df["shipping date (DateOrders)"]
    - df["order date (DateOrders)"]
).dt.total_seconds() / 86400

print("Calculated shipping duration:")
display(df["Calculated Shipping Days"].describe())

print("\nRecorded actual shipping days:")
display(df["Days for shipping (real)"].describe())

print("\nScheduled shipping days:")
display(df["Days for shipment (scheduled)"].describe())
```

```python
print("Recorded actual shipping days:")
display(df["Days for shipping (real)"].describe())

print("Scheduled shipping days:")
display(df["Days for shipment (scheduled)"].describe())

print("Calculated shipping days:")
display(df["Calculated Shipping Days"].describe())
```

```python
print(
    "Rows where calculated duration differs "
    "from recorded actual days:",
    (
        df["Calculated Shipping Days"]
        != df["Days for shipping (real)"]
    ).sum()
)
```

```python
display(df["Days for shipment (scheduled)"].describe())
```

```python
mismatch = (
    df["Calculated Shipping Days"].round(6)
    != df["Days for shipping (real)"]
)

print("Rows with differing durations:", mismatch.sum())
print("Percentage differing:", round(mismatch.mean() * 100, 2), "%")

display(
    df.loc[
        mismatch,
        [
            "order date (DateOrders)",
            "shipping date (DateOrders)",
            "Calculated Shipping Days",
            "Days for shipping (real)",
            "Days for shipment (scheduled)"
        ]
    ].head(10)
)
```

```python
df["Calculated Calendar Days"] = (
    df["shipping date (DateOrders)"].dt.normalize()
    - df["order date (DateOrders)"].dt.normalize()
).dt.days

calendar_mismatch = (
    df["Calculated Calendar Days"]
    != df["Days for shipping (real)"]
)

print(
    "Calendar-day mismatches:",
    calendar_mismatch.sum()
)

print(
    "Calendar-day mismatch percentage:",
    round(calendar_mismatch.mean() * 100, 2),
    "%"
)

display(
    df.loc[
        calendar_mismatch,
        [
            "order date (DateOrders)",
            "shipping date (DateOrders)",
            "Calculated Calendar Days",
            "Days for shipping (real)"
        ]
    ].head(10)
)
```

```python
risk_check = pd.crosstab(
    df["Delivery Status"],
    df["Late_delivery_risk"],
    margins=True
)

display(risk_check)
```

# Phase 2 — Dataset Acquisition and Exploration Summary

## Dataset Overview
- Dataset: DataCo SMART Supply Chain Dataset
- Total rows: 180,519
- Total columns: 53
- Data format: CSV
- Encoding: Latin-1

## Data Quality Findings
- `Product Description` is entirely missing.
- `Order Zipcode` has approximately 86.24% missing values.
- Eight customer last-name values and three customer ZIP-code values are missing.
- No duplicate complete rows were found.
- `Order Item Id` has no duplicate values.
- Both order and shipping date columns were successfully converted to datetime.
- Calendar-day shipping calculations matched `Days for shipping (real)` for all rows.
- All sales values are positive.
- There are 33,784 rows with negative profit, which require business interpretation rather than automatic deletion.

## Business Findings
- 98,977 rows are labeled `Late delivery` (approximately 54.83%).
- Standard Class accounts for the largest number of records.
- Delivery-status percentages differ across shipping modes.
- Average actual shipping time is approximately 3.50 days.

## Important Columns
Order IDs, order-item IDs, customer IDs, product IDs, sales, profit, order dates, shipping dates, shipping modes, delivery status, actual shipping days, and scheduled shipping days.

## Data Privacy
Customer emails and passwords must not be published or included in public reports, exports, or GitHub.
